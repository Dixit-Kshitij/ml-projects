```python
import numpy as np
from scipy.stats import norm
```

```python
sample = [172, 174, 168, 169, 171, 173, 175, 170, 169, 172]
u = 170
sigma = 3
xbar = np.mean(sample)
n = len(sample)
```

```python
z_score = (xbar-u)/(sigma/np.sqrt(n))
z_score
```

```python
alpha = 0.05
```

```python
p_value = 2*(1-norm.cdf(abs(z_score)))
p_value
```

```python
if p_value < alpha:
    print('reject null hypothesis')
else:
    print('accept null hypothesis')
```
