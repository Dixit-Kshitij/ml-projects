```python
from scipy import stats
import numpy as np
# Data for two groups
group_A = [85, 88, 90, 92, 87, 85, 89, 91, 86, 88]
group_B = [82, 84, 80, 83, 81, 79, 78, 85, 84, 83]

# Perform independent two-sample t-test (Welch’s t-test)
t_stat, p_value = stats.ttest_ind(group_A, group_B, equal_var=False)
n = len(group_A)
s1 = np.std(group_A)
s2 = np.std(group_B)
# Print results
print("T-statistic:", t_stat)
print("P-value:", p_value)

# Significance level
alpha = 0.05

# Hypothesis test decision
if p_value < alpha:
    print("Reject the null hypothesis")
    print("There is a statistically significant difference between the two groups.")
else:
    print("Fail to reject the null hypothesis")
    print("There is no statistically significant difference between the two groups.")
```
