```python
import seaborn as sns
import pandas as pd
from scipy.stats import f_oneway
```

```python
df = sns.load_dataset('titanic')
```

```python
df[['age','pclass']].isnull().sum()
```

```python
df = df[['age','pclass']].dropna()
```

```python
class_1 = df[df['pclass']==1]['age']
class_2 = df[df['pclass']==2]['age']
class_3 = df[df['pclass']==3]['age']
```

```python
df.head()
```

```python
f_stats,p_value = f_oneway(class_1,class_2,class_3)
```

```python
alpha = 0.05
```

```python
f_stats
```

```python
p_value
```

```python
if p_value < alpha:
    print('we reject null hypothesis thus significant relationship exists')
else:
    print('there is no significant difference')
```
