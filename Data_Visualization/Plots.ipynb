```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
y = np.linspace(1,5,11)
y
```

```python
x = y**2
x
```

```python
plt.subplot(2,2,1)
plt.plot(y)
plt.subplot(2,2,2)
plt.plot(x)
plt.subplot(2,2,3)
plt.plot(y**3)
plt.subplot(2,2,4)
plt.plot(x**4)
```

```python
fig = plt.figure()
axis1 = fig.add_axes([0.1,0.1,1.0,1.0])
axis1.plot(x,y)
axis2 = fig.add_axes([0.6,0.3,0.3,0.3])
axis2.plot(x,y)
# plt.show()
```

# Types of Plot

```python
fig = plt.figure(figsize=(2,4),dpi=1000)
axis1 = fig.add_axes([0.1,0.1,1.0,1.0])
axis1.plot(x,y)
axis2 = fig.add_axes([0.6,0.3,0.3,0.3])
axis2.plot(x,y)
# plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset('tips')
```

```python
df.shape
```

```python
df.info()
```

```python
df['size'].unique()
```

```python
sns.displot(df['total_bill'],bins = 5,kde = True)
```

```python
sns.histplot(df['tip'],bins = 5,kde = True)
```

```python
plt.subplot(1,2,1)
sns.histplot(df['total_bill'],bins = 20,kde = True)
plt.subplot(1,2,2)
sns.histplot(df['tip'],bins = 20,kde = True)
```

```python
sns.jointplot(x = 'tip',y = 'total_bill',data = df,kind='reg')
```

```python
sns.pairplot(df)
```

```python
sns.boxplot(hue='day',y='tip',data=df,palette='rainbow')
```

```python
sns.violinplot(hue='day',y='tip',data=df,palette='rainbow')
```

```python
sns.stripplot(x='day',y='tip',data=df)
```

```python
sns.swarmplot(x='day',y='tip',data=df)
```

```python
sns.violinplot(x='day',y='tip',data=df)
sns.swarmplot(x='day',y='tip',data=df)
```
