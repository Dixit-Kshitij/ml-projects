```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
tips = sns.load_dataset('tips')
```

```python
tips.head()
```

```python
numeric_col = ['total_bill','tip','size']
```

```python
sns.heatmap(tips[numeric_col].corr(),annot=True)
```

# Cluster map

```python
sns.clustermap(tips[numeric_col].corr(),annot=True)
```

# pivot table heat map

```python
flights = sns.load_dataset('flights')
```

```python
flights.head()
```

```python
pvtflight = flights.pivot_table(values = 'passengers',index = 'month',columns = 'year',observed=False,aggfunc='max')
```

```python
pvtflight
```

# Regression plot

```python
sns.lmplot(data=tips,x='total_bill',y='tip',hue='sex')
```
