```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('train.txt',sep=';',header = None,names = ['text','emotion'])
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
unique_emotion = df['emotion'].unique()
unique_emotion
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df['emotion'] = le.fit_transform(df['emotion'])
```

```python
df['emotion'].unique()
```

```python
df['text'] = df['text'].apply(lambda x: x.lower())
```

```python
import string 
def remove_punc(txt):
    return txt.translate(str.maketrans('','',string.punctuation))
df['text'] = df['text'].apply(remove_punc)
```

```python
def remove_numbers(txt):
    new = ""
    for i in txt:
        if not i.isdigit():
            new = new + i
    return new

df['text'] = df['text'].apply(remove_numbers)
```

```python
def remove_emojis(txt):
    new = ""
    for i in txt:
        if i.isascii():
            new += i
    return new

df['text'] = df['text'].apply(remove_emojis)
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
import nltk
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('punkt_tab')
```

```python
stop_words = set(stopwords.words('english'))
len(stop_words)
```

```python
type(stop_words)
```

```python
def remove(txt):
    words = word_tokenize(txt)
    cleaned = []
    for i in words:
        if i not in stop_words: 
            cleaned.append(i)
    return ' '.join(cleaned)

df['text'] = df['text'].apply(remove)
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(df['text'],df['emotion'],random_state=42,test_size=0.20)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer,CountVectorizer
```

```python
bow_vectorizer = CountVectorizer()
```

```python
X_train_bow = bow_vectorizer.fit_transform(X_train)
X_test_bow = bow_vectorizer.transform(X_test)
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
```

```python
nb = MultinomialNB()
```

```python
nb.fit(X_train_bow,y_train)
```

```python
y_pred = nb.predict(X_test_bow)
```

```python
accuracy = accuracy_score(y_test,y_pred)
accuracy
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem_words(text):
    return " ".join([ps.stem(word) for word in text.split()])
```

```python
sample = 'walk walking walked'
stem_words(sample)
```

```python
import nltk
from nltk.stem import WordNetLemmatizer
wordnet_lemmatizer = WordNetLemmatizer()

sentence = "He was running and eating at same time. He has bad habit of swimming after playing long hours in the Sun."
punctuations="?:!.,;"
sentence_words = nltk.word_tokenize(sentence)
for word in sentence_words:
     if word in punctuations:
         sentence_words.remove(word)

sentence_words
print("{0:20}{1:20}".format("Word","Lemma"))
for word in sentence_words:
    print ("{0:20}{1:20}".format(word,wordnet_lemmatizer.lemmatize(word,pos='v')))
```

```python
import numpy as np

def outlier(x):
    # x = np.array(x, dtype=int)

    Q1 = np.percentile(x, 25)
    Q3 = np.percentile(x, 75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = [v for v in x if (v < lower) or (v > upper)]
    
    if len(outliers) > 0:
        # print(" ".join(map(str, outliers)))
        print(*outliers)
```
